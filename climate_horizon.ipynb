###1. Introduction
1.1. Problem Statement
Weather forecasting is a crucial application of data science that impacts various sectors, including agriculture, energy, and tourism. Accurate temperature prediction helps in planning and decision-making. The goal of this project is to analyze historical temperature data and build a reliable time-series model to forecast daily temperatures.

###1.2. Objective
The primary objective is to develop a predictive model that can accurately forecast the daily temperature for a specific city for a user-defined number of future days.

###1.3. Dataset
The project utilizes a weather.csv dataset. The key columns used for this analysis are:

Date: The date of the observation.

Temperature: The recorded temperature in Celsius.

City: The city where the data was recorded.

##2. Methodology and Implementation
The project follows a structured approach from data preparation to model evaluation.

###2.1. Tools and Libraries Used 🧰
The analysis was performed using Python and the following essential libraries:

pandas: For data manipulation and loading the CSV file.

numpy: For numerical operations, especially for calculating metrics.

matplotlib: For data visualization to plot graphs and charts.

statsmodels: For implementing the ARIMA time-series model.

scikit-learn: For calculating model performance metrics.

###2.2. Data Loading and Preprocessing
The first step was to load the weather.csv dataset into a pandas DataFrame. The 'Date' column was converted from a string to a datetime object, which is essential for time-series analysis. This column was then set as the index of the DataFrame.

###2.3. Exploratory Data Analysis (EDA) 📊
Before modeling, an initial analysis was conducted to understand the data's structure and patterns.

Descriptive Statistics: Functions like describe() and info() were used to get a summary of the dataset, including mean, standard deviation, and data types.

Data Visualization:

A histogram of the 'Temperature' column was plotted to understand its distribution.

A time-series plot was created to visualize how the daily temperature changes over time, helping to identify trends or seasonality.

###2.4. Data Cleaning
The dataset was checked for missing values. To ensure data continuity without losing valuable information, any missing temperature values were handled using the forward-fill (ffill) method. This method propagates the last valid observation forward.

###2.5. Model Training: ARIMA
The ARIMA model was chosen for this forecasting task due to its effectiveness in handling time-series data. ARIMA stands for AutoRegressive Integrated Moving Average.

Train-Test Split: The historical data was split into a training set and a testing set. The last 7 days of data were reserved for the test set to evaluate the model's performance on unseen data.

Model Fitting: An ARIMA model with parameters (p=5, d=1, q=0) was trained on the training dataset. These parameters define the model's structure:

p=5: The number of lag observations (AutoRegressive part).

d=1: The number of times the raw observations are differenced (Integrated part).

q=0: The size of the moving average window (Moving Average part).

##3. Results and Evaluation
3.1. Forecasting
The trained ARIMA model was used to forecast the temperatures for the next 7 days, corresponding to the period covered by our test set.

###3.2. Performance Metrics 🎯
The model's predictions were compared against the actual temperatures from the test set. The following metrics were calculated:

Mean Absolute Error (MAE): The average absolute difference between predicted and actual values.

Root Mean Squared Error (RMSE): The square root of the average of squared differences. It penalizes larger errors more heavily.

R-squared (R2 Score): Indicates the proportion of the variance in the dependent variable that is predictable from the independent variable(s). A score closer to 1 indicates a better fit.

Forecast Accuracy: Calculated as 100 - MAPE (Mean Absolute Percentage Error), it gives an intuitive measure of the prediction accuracy.

```
#df[''].hist()
data['Temperature'].hist()
# Visualizing the time series
plt.figure(figsize=(10, 5))
plt.plot(data['Temperature'], label='Daily Temperature')
plt.title('Daily Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.show()
```

```
# Convert 'Date' to datetime and set as index for time series analysis
data['Date'] = pd.to_datetime(data['Date'], dayfirst=True)
data.set_index('Date', inplace=True)
# # 4. (Optional) Filter for a specific city if dataset has multiple cities
city = data['City'].iloc[0]
city_data = data[data['City'] == city]
# 5. Visualize the temperature time series
plt.figure(figsize=(10, 5))
plt.plot(city_data.index, city_data['Temperature'], marker='o', label=f"Temperature in {city}")
plt.title(f"Daily Temperature in {city}")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.legend()
plt.show()
# 6. Check for missing values and handle them if needed
print(city_data.isnull().sum())
city_data['Temperature'].fillna(method='ffill', inplace=True)  # Forward fill for missing values
```

```
# 7. Split data into train and test sets (e.g., last 7 days for testing)
train = city_data['Temperature'][:-7]
test = city_data['Temperature'][-7:]
```

```
# 9. Train the ARIMA model
model = ARIMA(train, order=(p, d, q))
model_fit = model.fit()
```

```
# 10. Forecast for the next 7 days
forecast = model_fit.forecast(steps=7)
print(f"Predicted temperatures for next 7  days:\n", forecast)
```

```
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Calculate Mean Absolute Error (MAE)
mae = mean_absolute_error(test, forecast)
print(f"Mean Absolute Error (MAE): {mae:.2f}")

# Calculate Mean Squared Error (MSE)
mse = mean_squared_error(test, forecast)
print(f"Mean Squared Error (MSE): {mse:.2f}")

# Calculate Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

# Calculate R-squared (Coefficient of Determination)
r2 = r2_score(test, forecast)
print(f"R-squared (R2 Score): {r2:.2f}")
# Calculate MAPE -  Mean Absolute Percentage Error (MAPE)
mape = np.mean(np.abs((test.values - forecast) / test.values)) * 100
print(f"MAPE: {mape:.2f}%")

# Calculate Accuracy
accuracy = 100 - mape
print(f"Forecast Accuracy: {accuracy:.2f}%")
# Accuracy analysis (for regression, accuracy is not a direct metric like classification,
# but you can analyze how close predictions are to actual values)
# Let's also print a comparison table for actual vs predicted
comparison = pd.DataFrame({
    'Actual': test.values,
    'Predicted': forecast
}, index=test.index)
print("\nActual vs Predicted Temperatures:")
print(comparison)
```

```
# 11. Visualize actual vs predicted temperatures
plt.figure(figsize=(8, 4))
plt.plot(test.index, test.values, marker='o', label='Actual')
plt.plot(test.index, forecast, marker='x', linestyle='--', label='Predicted')
plt.title(f'Actual vs Predicted Temperatures in {city}')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True)
plt.show()
```

```
# 9. Forecast for user-defined days
days_to_predict = int(input("Enter the number of days to predict: "))

forecast = model_fit.forecast(steps=days_to_predict)
print(f"Predicted temperatures for next {days_to_predict} days:\n", forecast)
```
